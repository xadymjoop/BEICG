# Exercice 1 : Mise en place d'un Calandrier

```python
def afficher_calendrier():
    # Saisie des données utilisateur
    nb_jours = int(input("Entrez le nombre de jours dans le mois : "))
    premier_jour = int(input("Entrez le numéro du jour de début du mois (1 = lundi, ..., 7 = dimanche) : "))

    # Vérification des entrées
    if nb_jours < 28 or nb_jours > 31:
        print("Le nombre de jours doit être compris entre 28 et 31.")
        return
    if premier_jour < 1 or premier_jour > 7:
        print("Le numéro du premier jour doit être entre 1 (lundi) et 7 (dimanche).")
        return

    # Entête du calendrier
    jours = ["LUN", "MAR", "MER", "JEU", "VEN", "SAM", "DIM"]
    print(" ".join(jours))

    # Calcul des espaces avant le premier jour
    espace_initial = (premier_jour - 1) * 4
    print(" " * espace_initial, end="")

    # Impression des jours
    for jour in range(1, nb_jours + 1):
        print(f"{jour:2} ", end="")
        # Si c'est le dernier jour de la semaine, passer à la ligne suivante
        if (jour + premier_jour - 1) % 7 == 0:
            print()
    print()


#  Exécuter le programme
afficher_calendrier()
```

```python
import random

# Base de données : les 31 régions de la Côte d'Ivoire et leurs chefs-lieux
regions = {
    "Agnéby-Tiassa": "Agboville",
    "Bafing": "Touba",
    "Bagoué": "Boundiali",
    "Belier": "Yamoussoukro",
    "Béré": "Mankono",
    "Bounkani": "Bouna",
    "Cavally": "Guiglo",
    "Folon": "Minignan",
    "Gbêkê": "Bouaké",
    "Gbôklè": "Sassandra",
    "Gôh": "Gagnoa",
    "Gontougo": "Bondoukou",
    "Grands Ponts": "Dabou",
    "Guémon": "Duékoué",
    "Hambol": "Katiola",
    "Indénié-Djuablin": "Abengourou",
    "Kabadougou": "Odienné",
    "Lôh-Djiboua": "Divo",
    "Marahoué": "Bouaflé",
    "Mé": "Adzopé",
    "Moronou": "Bongouanou",
    "N'Zi": "Dimbokro",
    "Poro": "Korhogo",
    "San-Pédro": "San-Pédro",
    "Sud-Comoé": "Aboisso",
    "Tchologo": "Ferkessédougou",
    "Tonkpi": "Man",
    "Worodougou": "Séguéla",
    "Yamoussoukro": "Yamoussoukro",
    "Bélier": "Toumodi",
    "Zanzan": "Bondoukou"
}

# Liste pour stocker les meilleurs scores
meilleurs_scores = []

# Fonction pour afficher les meilleurs scores
def afficher_meilleurs_scores():
    print("\n=== Meilleurs Scores ===")
    meilleurs_scores_tries = sorted(meilleurs_scores, key=lambda x: x[1], reverse=True)[:5]
    for i, (pseudo, score) in enumerate(meilleurs_scores_tries, 1):
        print(f"{i}. {pseudo} - {score} points")
    print("========================\n")

# Fonction pour jouer une partie
def jouer_partie():
    # Afficher les meilleurs scores au début
    afficher_meilleurs_scores()

    # Mélanger les régions pour éviter les répétitions
    questions = random.sample(list(regions.items()), 10)
    score = 0



    print("=== Début de la partie ===")
    for region, chef_lieu in questions:
        # Poser la question
        reponse = input(f"Quel est le chef-lieu de la région {region}? ")

        # Vérifier la réponse
        if reponse.strip().lower() == chef_lieu.lower():
            print("Bonne réponse.")
            score += 10  # Ajouter des points pour une bonne réponse
        else:
            print(f"Réponse incorrecte! Le chef-lieu de {region} est {chef_lieu}.")

    # Calcul du score final
    print(f"\nFin du jeu, vous avez obtenu {score} points sur 100.")
    pseudo = input("Entrez votre pseudo pour sauvegarder votre score : ")
    meilleurs_scores.append((pseudo, score))

    # Proposer une nouvelle partie
    rejouer = input("Voulez-vous effectuer une autre partie ? [o/n] ").strip().lower()
    if rejouer == 'o':
        jouer_partie()
    else:
        print("OKAY BYE BYE!!!")

# Lancer le jeu
jouer_partie()
```
